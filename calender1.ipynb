### Calculating New Moon Dates using `ephem`

To align our Lakota calendar more closely with actual lunar cycles, we can use the `ephem` library to find the dates of the new moon. In many traditional lunar calendars, the new moon signifies the beginning of a new month.

`ephem` allows us to precisely calculate the phase of the moon. We'll look for `ephem.next_new_moon()` to find the exact moment a new moon occurs. Since the new moon can occur at any time of day, and a 'new month' traditionally begins at sunset following the new moon, or simply on the day of the new moon, we'll focus on the date of the new moon as our starting point.

```python
def get_new_moon_dates(start_year, end_year):
    """
    Calculates the dates of all new moons within a specified year range.

    Args:
        start_year (int): The starting year (inclusive).
        end_year (int): The ending year (inclusive).

    Returns:
        list: A list of datetime.date objects for each new moon.
    """
    new_moons = []
    current_date = datetime.date(start_year, 1, 1)

    while current_date.year <= end_year:
        # Find the next new moon from the current date
        next_nm_ephem = ephem.next_new_moon(current_date.strftime('%Y/%m/%d %H:%M:%S'))
        # Convert the ephem date to UTC datetime object
        next_nm_utc = ephem.Date(next_nm_ephem).datetime()
        # Note: ephem calculates in UTC, so we will use the UTC date for now.
        new_moons.append(next_nm_utc.date())

        # Advance current_date to the day after the found new moon to search for the next one
        current_date = next_nm_utc.date() + datetime.timedelta(days=1)

    return new_moons

# Example: Get new moon dates for a specific year
print("New Moon Dates for 2024:")
new_moons_2024 = get_new_moon_dates(2024, 2024)
for nm_date in new_moons_2024:
    print(f"- {nm_date.strftime('%Y-%m-%d')}")

print("\nNew Moon Dates for 2025:")
new_moons_2025 = get_new_moon_dates(2025, 2025)
for nm_date in new_moons_2025:
    print(f"- {nm_date.strftime('%Y-%m-%d')}")
```

### Accurate Lakota Calendar Mapping using `ephem`

To accurately map Gregorian dates to Lakota moon months, we need to consider two key astronomical events:

1.  **Spring Equinox:** The Lakota New Year is traditionally marked by the New Moon following the Spring Equinox.
2.  **New Moons:** Each Lakota month begins with a new moon.

We will first implement a function to find the Spring Equinox, then use that to establish the start of the Lakota New Year, and subsequently determine the boundaries for all 13 (or 14) months in a given Lakota year. The 13th month, `Wicanhcala Wi`, is an intercalary month inserted to keep the lunar calendar aligned with the solar year, typically when there are 13 new moons between consecutive Spring Equinox new moons.

```python
def get_spring_equinox_date(year):
    """
    Calculates the date of the Spring Equinox for a given year.

    Args:
        year (int): The year.

    Returns:
        datetime.date: The date of the Spring Equinox.
    """
    # Create an ephem observer at the Earth's center (no specific location needed for equinox)
    obs = ephem.Observer()
    obs.date = f'{year}/3/20'

    # Find the next vernal equinox. ephem.next_equinox returns the date/time
    # of the next equinox from obs.date, so we might need to adjust if it's for the previous year.
    equinox_date_ephem = ephem.next_equinox(obs.date)

    # Check if the calculated equinox is for the target year or the next
    equinox_datetime_utc = ephem.Date(equinox_date_ephem).datetime()
    if equinox_datetime_utc.year > year:
        # If it's for the next year, search backwards from the year's end
        obs.date = f'{year}/12/31'
        equinox_date_ephem = ephem.previous_equinox(obs.date)
        equinox_datetime_utc = ephem.Date(equinox_date_ephem).datetime()

    return equinox_datetime_utc.date()

# Example usage:
spring_equinox_2024 = get_spring_equinox_date(2024)
print(f"Spring Equinox 2024: {spring_equinox_2024}")

spring_equinox_2025 = get_spring_equinox_date(2025)
print(f"Spring Equinox 2025: {spring_equinox_2025}")
```

```python
def get_summer_solstice_date(year):
    """
    Calculates the date of the Summer Solstice for a given year.

    Args:
        year (int): The year.

    Returns:
        datetime.date: The date of the Summer Solstice.
    """
    # Create an ephem observer
    obs = ephem.Observer()
    # Set date to around the Summer Solstice
    obs.date = f'{year}/6/20'

    # Find the next solstice. ephem.next_solstice returns the date/time
    solstice_date_ephem = ephem.next_solstice(obs.date)

    # Convert the ephem date to UTC datetime object
    solstice_datetime_utc = ephem.Date(solstice_date_ephem).datetime()

    # Check if the calculated solstice is for the target year or the next
    if solstice_datetime_utc.year > year:
        # If it's for the next year, search backwards from the year's end
        obs.date = f'{year}/12/31'
        solstice_date_ephem = ephem.previous_solstice(obs.date)
        solstice_datetime_utc = ephem.Date(solstice_date_ephem).datetime()

    return solstice_datetime_utc.date()

# Example usage:
summer_solstice_2028 = get_summer_solstice_date(2028)
print(f"Summer Solstice 2028: {summer_solstice_2028}")
```

### Determining the Lakota New Year and Monthly Boundaries

With the Spring Equinox and new moon dates, we can define the Lakota New Year. The first new moon following the Spring Equinox marks the beginning of the Lakota calendar cycle for that year. From this point, we can assign the 12 primary Lakota moon names sequentially. The 13th month, `Wicanhcala Wi` (Old Woman Moon), is inserted as an intercalary month when a particular Lakota year contains 13 new moons between consecutive Lakota New Year moons.

```python
def get_lakota_year_moons(year):
    """
    Generates a list of Lakota moon months for a given year, based on new moon dates
    and the Spring Equinox.

    Args:
        year (int): The Gregorian year for which to generate the Lakota calendar.

    Returns:
        list: A list of dictionaries, each representing a Lakota month with its name,
              English meaning, and start/end dates. Includes the intercalary month if present.
    """
    lakota_year_calendar = []

    # 1. Get Spring Equinox for the current and next year to define the Lakota year bounds
    spring_equinox_current_year = get_spring_equinox_date(year)
    spring_equinox_next_year = get_spring_equinox_date(year + 1)

    # 2. Get all new moons around the relevant period
    # We need new moons from a bit before the current year's equinox up to a bit after the next year's equinox
    all_new_moons = get_new_moon_dates(year - 1, year + 2)
    # Filter to get only unique dates, as get_new_moon_dates might return duplicates around year boundaries
    all_new_moons = sorted(list(set(all_new_moons)))

    # 3. Find the Lakota New Year (first new moon AFTER the current year's Spring Equinox)
    lakota_new_year_start = None
    for nm_date in all_new_moons:
        if nm_date >= spring_equinox_current_year:
            lakota_new_year_start = nm_date
            break

    if not lakota_new_year_start:
        # This should ideally not happen if the date range for all_new_moons is sufficient
        raise ValueError(f"Could not find Lakota New Year start for {year}")

    # 4. Determine the end of the Lakota year (first new moon AFTER the next year's Spring Equinox)
    lakota_next_year_start = None
    found_first_new_moon_after_next_equinox = False
    for nm_date in all_new_moons:
        if nm_date >= spring_equinox_next_year:
            if not found_first_new_moon_after_next_equinox:
                lakota_next_year_start = nm_date
                found_first_new_moon_after_next_equinox = True
            # We need to find the one after the next Spring Equinox
            # But if there are *two* new moons before the next Spring Equinox new moon
            # AND the first of them is after the *current* Spring Equinox, we need to make sure
            # we capture all months between them. This loop structure needs refinement.
            # Let's simplify: find all new moons between lakota_new_year_start and lakota_next_year_start.

    if not lakota_next_year_start:
        raise ValueError(f"Could not find Lakota Next Year start for {year}")

    # Extract new moons for *this* Lakota year
    new_moons_for_lakota_year = []
    for nm_date in all_new_moons:
        if lakota_new_year_start <= nm_date < lakota_next_year_start:
            new_moons_for_lakota_year.append(nm_date)

    # Add the Lakota New Year start if it's not already in the list (it should be)
    if lakota_new_year_start not in new_moons_for_lakota_year:
        new_moons_for_lakota_year.insert(0, lakota_new_year_start)

    # Ensure sorted and unique (should already be due to earlier sorting and filtering)
    new_moons_for_lakota_year = sorted(list(set(new_moons_for_lakota_year)))

    # Assign Lakota month names
    num_lakota_months = len(lakota_months) # Should be 13, including the intercalary one at the end
    num_new_moons_in_year = len(new_moons_for_lakota_year)

    # Determine if it's a 13-moon year or a 12-moon year (with an inserted 13th month)
    # The traditional 13-moon calendar has 12 main months and an occasional 13th intercalary.
    # So, if we have 13 new moons in the cycle, the last one will be the Wicanhcala Wi.
    # If we have 12 new moons, there is no Wicanhcala Wi. This implies the 'lakota_months' list
    # will have one unused entry sometimes.

    # Let's redefine the Lakota month names without the intercalary month in the base list,
    # and add it only if there are 13 new moons.
    primary_lakota_months = lakota_months[:12] # First 12 months
    wicanhcala_wi = lakota_months[12] # The 13th, intercalary month

    effective_lakota_months_for_year = list(primary_lakota_months)
    if num_new_moons_in_year > 12: # This means there are 13 new moons in this Lakota year
        effective_lakota_months_for_year.append(wicanhcala_wi)
    elif num_new_moons_in_year < 12:
         # This case indicates an error in the logic or an unusual astronomical event
         # For a precise calendar, we expect 12 or 13 new moons per cycle.
         print(f"Warning: Found only {num_new_moons_in_year} new moons for Lakota year starting {lakota_new_year_start}")

    # Create the calendar structure
    for i in range(len(new_moons_for_lakota_year)):
        month_start_date = new_moons_for_lakota_year[i]
        # The end date of a moon month is the day before the next new moon
        if i + 1 < len(new_moons_for_lakota_year):
            month_end_date = new_moons_for_lakota_year[i+1] - datetime.timedelta(days=1)
        else:
            # If this is the last new moon in our list, its end date is just before the *next* Lakota New Year start.
            # This implies the last month might extend into the Gregorian 'next year'
            month_end_date = lakota_next_year_start - datetime.timedelta(days=1)

        if i < len(effective_lakota_months_for_year):
            lakota_name, english_name = effective_lakota_months_for_year[i]
        else:
            # Fallback if somehow more new moons than defined month names
            lakota_name, english_name = ("Unknown Moon", "Unknown Moon")

        lakota_year_calendar.append({
            "lakota_name": lakota_name,
            "english_name": english_name,
            "start_date": month_start_date,
            "end_date": month_end_date
        })

    return lakota_year_calendar

# Example usage:
print("\nLakota Calendar for 2024 (based on New Year starting after Spring Equinox):")
lakota_2024_calendar = get_lakota_year_moons(2024)
for month in lakota_2024_calendar:
    print(f"- {month['lakota_name']} ({month['english_name']}): {month['start_date']} to {month['end_date']}")
```

### Mapping a Gregorian Date to a Lakota Moon Month

Now we can create a function that takes any Gregorian date and, using the generated Lakota calendar for that year, returns the corresponding Lakota moon month. This function will iterate through the calendar entries and find the month whose date range encompasses the input Gregorian date.

```python
def get_lakota_moon_month_for_date(gregorian_date):
    """
    Maps a given Gregorian date to its corresponding Lakota moon month.

    Args:
        gregorian_date (datetime.date): The Gregorian date to map.

    Returns:
        dict or None: A dictionary containing the Lakota month's details (name, English meaning, start/end dates),
                      or None if the date does not fall within a defined Lakota month for that year.
    """
    # Determine the Gregorian year that the Lakota calendar cycle starts in for the given date.
    # This is slightly complex because a Lakota year can span across Gregorian year boundaries.
    # For simplicity, we'll try to generate the calendar for the Gregorian year of the input date,
    # and one year prior, as the Lakota New Year could start late in the previous Gregorian year.

    # First, let's determine the relevant Lakota year. A Lakota year starts with a new moon after the Spring Equinox.
    # If the gregorian_date is before the current Gregorian year's Spring Equinox,
    # it might belong to the Lakota year that started in the *previous* Gregorian year.

    # Get the Spring Equinox for the input date's Gregorian year
    spring_equinox = get_spring_equinox_date(gregorian_date.year)

    lakota_year_to_check = gregorian_date.year
    if gregorian_date < spring_equinox:
        # If the date is before the Spring Equinox, it could be part of the Lakota year
        # that started in the previous Gregorian year.
        # However, the `get_lakota_year_moons` function already handles spanning Gregorian years
        # by finding all new moons between two consecutive Lakota New Year starts.
        # So, we just need to generate for the Gregorian year the date falls into,
        # and the function will correctly capture its Lakota year.
        pass # No change needed to lakota_year_to_check for now.

    lakota_calendar = get_lakota_year_moons(lakota_year_to_check)

    # Iterate through the months in the generated calendar to find the match
    for month_data in lakota_calendar:
        if month_data["start_date"] <= gregorian_date <= month_data["end_date"]:
            return month_data

    # If not found in the current Gregorian year's Lakota calendar, try the previous one
    # This covers cases where the Lakota year starts in the previous Gregorian year
    # and extends into the current one, but the input date is early in the current Gregorian year.
    lakota_calendar_prev_year = get_lakota_year_moons(lakota_year_to_check - 1)
    for month_data in lakota_calendar_prev_year:
        if month_data["start_date"] <= gregorian_date <= month_data["end_date"]:
            return month_data

    return None

# Example usage:
print("\nMapping Gregorian Dates to Lakota Moon Months:")

today = datetime.date.today()
lakota_month_today = get_lakota_moon_month_for_date(today)
if lakota_month_today:
    print(f"Today ({today}): {lakota_month_today['lakota_name']} ({lakota_month_today['english_name']})")
else:
    print(f"Today ({today}): No Lakota month found.")

example_date_early_year = datetime.date(2024, 1, 15) # Example early in the Gregorian year
lakota_month_early = get_lakota_moon_month_for_date(example_date_early_year)
if lakota_month_early:
    print(f"{example_date_early_year}: {lakota_month_early['lakota_name']} ({lakota_month_early['english_name']})")
else:
    print(f"{example_date_early_year}: No Lakota month found.")

example_date_late_year = datetime.date(2024, 12, 20) # Example late in the Gregorian year
lakota_month_late = get_lakota_moon_month_for_date(example_date_late_year)
if lakota_month_late:
    print(f"{example_date_late_year}: {lakota_month_late['lakota_name']} ({lakota_month_late['english_name']})")
else:
    print(f"{example_date_late_year}: No Lakota month found.")

example_date_next_gregorian_year = datetime.date(2025, 2, 10) # Example in next Gregorian year, but same Lakota year
lakota_month_next_greg = get_lakota_moon_month_for_date(example_date_next_gregorian_year)
if lakota_month_next_greg:
    print(f"{example_date_next_gregorian_year}: {lakota_month_next_greg['lakota_name']} ({lakota_month_next_greg['english_name']})")
else:
    print(f"{example_date_next_gregorian_year}: No Lakota month found.")
```

### Identifying Intercalary Years

An intercalary year in the Lakota calendar is one where a 13th moon, `Wicanhcala Wi` (Old Woman Moon), is inserted to keep the lunar calendar aligned with the solar year. This typically occurs when there are 13 new moons between consecutive Lakota New Year starts (which begin with the new moon following the Spring Equinox).

```python
current_gregorian_year = datetime.date.today().year
next_decade_start_year = current_gregorian_year + 1
next_decade_end_year = current_gregorian_year + 10

intercalary_years = []

print(f"\nUpcoming Intercalary Years ({next_decade_start_year}-{next_decade_end_year}):")

for year in range(next_decade_start_year, next_decade_end_year + 1):
    try:
        # Get the Lakota calendar for the current Gregorian year
        lakota_calendar_for_year = get_lakota_year_moons(year)

        # If the calendar contains 13 months, it's an intercalary year
        if len(lakota_calendar_for_year) == 13:
            intercalary_years.append(year)

    except ValueError as e:
        print(f"Could not process year {year}: {e}")

if intercalary_years:
    for year in intercalary_years:
        print(f"- {year}")
else:
    print("No intercalary years found in the next decade based on the current calculation logic.")
```

### Function to Add Multiple Markers from a List

This helper function will take a `folium` map object and a list of site dictionaries, adding each site as a marker to the map.

```python
import folium # Ensure folium is imported if this cell is run independently

def add_markers_from_list(map_obj, sites_list):
    """
    Adds multiple markers to a folium map from a list of site dictionaries.

    Args:
        map_obj (folium.Map): The folium map object to add markers to.
        sites_list (list): A list of dictionaries, where each dictionary represents
                           a site with at least 'name', 'latitude', and 'longitude'.
                           Optional keys: 'icon_color', 'icon_type'.
    """
    for site in sites_list:
        name = site.get('name', 'Unknown Site')
        latitude = site.get('latitude')
        longitude = site.get('longitude')
        icon_color = site.get('icon_color', 'blue') # Default color
        icon_type = site.get('icon_type', 'info-sign') # Default icon

        if latitude is not None and longitude is not None:
            folium.Marker(
                location=[latitude, longitude],
                popup=name,
                icon=folium.Icon(color=icon_color, icon=icon_type)
            ).add_to(map_obj)
        else:
            print(f"Warning: Skipping site '{name}' due to missing coordinates.")

    return map_obj

print("Function 'add_markers_from_list' created successfully.")
```
