```python
import random
import time

# ==========================================
# 1. GLOBAL GAME CONFIGURATION & DATABASE
# ==========================================
TRUNK_CAPACITY = 100
wallet = 150.00
fuel_level = 20 # Initial fuel level in gallons
FUEL_CONSUMPTION_PER_DAY = 2 # Gallons consumed per day
GASOLINE_PRICE_PER_GALLON = 3.50

# Complete modern conversion grid list matching your exact inventory properties
CARGO_CATALOG = {
    "Commods (Govt Cheese)": {"cost": 18.00, "space": "2x2", "units": 4, "max_stack": 25,  "qty": 0},
    "Flour Sack":         {"cost": 8.00,  "space": "1x2", "units": 2, "max_stack": 5,   "qty": 0},
    "Jerked Meat":        {"cost": 15.00, "space": "2x2", "units": 4, "max_stack": 150, "qty": 0},
    "Gravel Road Map":    {"cost": 5.00,  "space": "1x1", "units": 1, "max_stack": 1,   "qty": 1},
    "Harmonica":          {"cost": 10.00, "space": "1x1", "units": 1, "max_stack": 1,   "qty": 1},
    "Spare Alternator":   {"cost": 50.00, "space": "2x2", "units": 4, "max_stack": 2,   "qty": 0},
    "Emergency Coolant":  {"cost": 15.00, "space": "1x2", "units": 2, "max_stack": 3,   "qty": 0},
    "Mechanic Toolbox":   {"cost": 45.00, "space": "2x3", "units": 6, "max_stack": 1,   "qty": 0},
    "First Aid Kit":      {"cost": 20.00, "space": "1x2", "units": 2, "max_stack": 3,   "qty": 1},
    "Heavy Blanket":      {"cost": 12.00, "space": "2x2", "units": 4, "max_stack": 5,   "qty": 2},
    "Pellet Rifle Case":  {"cost": 55.00, "space": "1x4", "units": 4, "max_stack": 1,   "qty": 0},
    "Lead Pellets (Box)": {"cost": 6.50,  "space": "1x1", "units": 1, "max_stack": 250, "qty": 50}
}

def intro():
    print("=================================================================")
    print("      THE RESERVATION TRAIL: MECHANICAL BREAKDOWN SCENARIOS     ")
    print("=================================================================")
    print("You start your journey with exactly $150.00 in cash.")
    print("PHASE 1: Drive to Rapid City, SD, to buy supplies without bursting your trunk grid.")
    print("PHASE 2: Face thunderstorms, tornadoes, blizzards, and deadzones to get home.")
    print("Manage your under-the-hood components closely to prevent total breakdown!")
    print()
    print("...") # Placeholder for game story
    time.sleep(2)

def calculate_current_trunk_space():
    total_space_used = 0
    for item_name, details in CARGO_CATALOG.items():
        if details["qty"] > 0:
            # Assuming 'space' is like '1x1' or '2x2', we need to calculate units.
            # However, the CARGO_CATALOG already has 'units' defined, so we use that.
            total_space_used += details['qty'] * details['units']
    return total_space_used

def market_price_fluctuation():
    global CARGO_CATALOG
    changed_prices = {}
    for item, details in CARGO_CATALOG.items():
        if item not in ["Gravel Road Map", "Harmonica", "First Aid Kit", "Mechanic Toolbox", "Pellet Rifle Case"]: # Items that might not fluctuate as much
            # Randomly adjust price by -10% to +10%
            change_percent = random.uniform(-0.10, 0.10)
            new_cost = details["cost"] * (1 + change_percent)
            # Ensure price doesn't go below $1.00
            details["cost"] = round(max(new_cost, 1.00), 2)
            if change_percent != 0: # Only record if there was an actual change
                changed_prices[item] = change_percent * 100

    if changed_prices:
        print("\n--- Market Price Fluctuations ---")
        for item, percent in changed_prices.items():
            print(f"{item}: {'Increased' if percent > 0 else 'Decreased'} by {abs(percent):.2f}%")
        print("-----------------------------------")
    else:
        print("\n--- Market prices remained stable. ---")

def rapid_city_market():
    global wallet, TRUNK_CAPACITY, CARGO_CATALOG, fuel_level, GASOLINE_PRICE_PER_GALLON
    print("\n🏙️ [ MILESTONE: ARRIVED IN RAPID CITY, SD ]")
    print("You pull into town to load your trunk before tackling the reservation highways.")
    market_price_fluctuation()

    while True:
        # Display current status
        print("\n=======================================================")
        used_space = calculate_current_trunk_space()
        print(f"💰 WALLET: ${wallet:.2f}  |  📦 TRUNK SPACE: {used_space} / {TRUNK_CAPACITY} Slots | ⛽ FUEL: {fuel_level:.1f} gal | 🧀 CHEESE: {CARGO_CATALOG['Commods (Govt Cheese)']['qty']}")
        print("=======================================================")
        print("Item Name            | Qty   | Grid Space | Max/Stack ")
        print("-------------------------------------------------------")
        for item, details in CARGO_CATALOG.items():
            if details["qty"] > 0:
                print(f"{item:<20} | {details['qty']:<5} | {details['space']:<10} | Up to {details['max_stack']}")
        print("-------------------------------------------------------")

        print("\n🛒 [ RAPID CITY MARKET CATALOG ]")
        # Display items for sale
        market_items = [item for item in CARGO_CATALOG if item != "Gravel Road Map"] # Assuming map isn't sold here
        for i, item in enumerate(market_items):
            cost = CARGO_CATALOG[item]["cost"]
            space = CARGO_CATALOG[item]["space"]
            print(f"{i+1}. {item:<20} (${cost:.2f} | Size: {space})")

        # Add Gasoline option
        print(f"{len(market_items) + 1}. Buy Gasoline (${GASOLINE_PRICE_PER_GALLON:.2f} / gallon)")
        gasoline_option_num = len(market_items) + 1

        # Add Sell Item option
        print(f"{len(market_items) + 2}. Sell Item")
        sell_option_num = len(market_items) + 2

        # Add Leave Town option
        print(f"{len(market_items) + 3}. Secure Cargo & Leave Town")
        leave_option_num = len(market_items) + 3


        choice_input = input("\nWhat would you like to do? (Enter number): ").strip()

        if not choice_input.isdigit():
            print("Invalid input. Please enter a number.")
            continue

        choice = int(choice_input)

        if choice == leave_option_num:
            print("Securing cargo and leaving Rapid City...")
            break
        elif choice == gasoline_option_num:
            while True:
                buy_gallons_input = input(f"How many gallons of gasoline would you like to buy (${GASOLINE_PRICE_PER_GALLON:.2f} per gallon)? ").strip()
                if not buy_gallons_input.isdigit():
                    print("Invalid input. Please enter a whole number for gallons.")
                    continue
                buy_gallons = int(buy_gallons_input)

                if buy_gallons <= 0:
                    print("You must buy at least one gallon.")
                    continue

                cost_of_gas = buy_gallons * GASOLINE_PRICE_PER_GALLON
                if wallet >= cost_of_gas:
                    wallet -= cost_of_gas
                    fuel_level += buy_gallons
                    print(f"Bought {buy_gallons} gallons of gasoline for ${cost_of_gas:.2f}. Fuel level: {fuel_level:.1f} gallons.")
                    break
                else:
                    print(f"Not enough money. You need ${cost_of_gas:.2f} but have ${wallet:.2f}.")
                    break

        elif choice == sell_option_num:
            print("\n--- Your Inventory to Sell ---")
            sellable_items = {k: v for k, v in CARGO_CATALOG.items() if v['qty'] > 0}
            if not sellable_items:
                print("You have no items to sell.")
                continue
            for i, (item_name, item_details) in enumerate(sellable_items.items()):
                print(f"{i+1}. {item_name} (Qty: {item_details['qty']}) - Est. Sale Value: ${item_details['cost'] * 0.75:.2f}")
            print(f"{len(sellable_items) + 1}. Back to Market")

            sell_choice_input = input("Choose an item to sell or 'Back to Market': ").strip()
            if not sell_choice_input.isdigit():
                print("Invalid input.")
                continue
            sell_choice = int(sell_choice_input)
            if sell_choice == len(sellable_items) + 1:
                continue

            if 1 <= sell_choice <= len(sellable_items):
                item_to_sell_name = list(sellable_items.keys())[sell_choice - 1]
                item_details = CARGO_CATALOG[item_to_sell_name]
                sell_qty_input = input(f"How many {item_to_sell_name} would you like to sell (max {item_details['qty']})? ").strip()
                if not sell_qty_input.isdigit():
                    print("Invalid quantity.")
                    continue
                sell_qty = int(sell_qty_input)
                if 0 < sell_qty <= item_details['qty']:
                    sale_value = item_details['cost'] * 0.75 * sell_qty
                    wallet += sale_value
                    item_details['qty'] -= sell_qty
                    print(f"Sold {sell_qty} {item_to_sell_name} for ${sale_value:.2f}. Wallet: ${wallet:.2f}")
                else:
                    print("Invalid quantity.")
            else:
                print("Invalid choice.")
        elif 1 <= choice <= len(market_items):
            selected_item_name = market_items[choice - 1]
            selected_item = CARGO_CATALOG[selected_item_name]

            # Check for max stack limit before buying
            current_qty = selected_item['qty']
            max_stack = selected_item['max_stack']

            if current_qty >= max_stack:
                print(f"You already have the maximum stack of {max_stack} for {selected_item_name}. Cannot buy more.")
                continue

            buy_qty_input = input(f"How many {selected_item_name} would you like to buy (max {max_stack - current_qty})? ").strip()
            if not buy_qty_input.isdigit():
                print("Invalid input. Please enter a number.")
                continue
            buy_qty = int(buy_qty_input)

            if buy_qty <= 0:
                print("Quantity must be positive.")
                continue

            if current_qty + buy_qty > max_stack:
                print(f"You can only buy {max_stack - current_qty} more {selected_item_name} to respect the max stack limit of {max_stack}.")
                continue

            total_cost = selected_item["cost"] * buy_qty
            item_space = selected_item["units"] * buy_qty

            if wallet >= total_cost:
                if calculate_current_trunk_space() + item_space <= TRUNK_CAPACITY:
                    wallet -= total_cost
                    selected_item["qty"] += buy_qty
                    print(f"Bought {buy_qty} {selected_item_name} for ${total_cost:.2f}.")
                else:
                    print("Not enough trunk space.")
            else:
                print("Not enough money.")
        else:
            print("Invalid choice. Please try again.")

def handle_snowstorm():
    global fuel_level # Declare global to modify fuel_level
    print("\n🌨️ Snowstorm ahead! Visibility is low. Driving becomes more difficult and consumes extra fuel.")

    # Consequence 1: Increased fuel consumption
    extra_fuel_consumed = 1 # Extra gallon for snowstorm
    fuel_level -= extra_fuel_consumed
    print(f"Due to heavy snow, an extra {extra_fuel_consumed} gallon of fuel was consumed.")

    # Consequence 2: Reduced visibility (narrative for now, could be integrated into mechanics later)
    print("Your progress for the day is slower than usual due to the poor conditions.")

    # Consequence 3: Small chance of engine trouble
    if random.random() < 0.15: # 15% chance of engine trouble
        print("The cold and strain on the engine might be causing issues...")
        handle_engine_trouble()

def handle_thunderstorm():
    global fuel_level # Declare global to modify fuel_level
    print("\n⛈️ Thunderstorm! Expect slippery roads, potential lightning, and increased strain on your vehicle.")

    # Consequence 1: Increased fuel consumption
    extra_fuel_consumed = 0.5 # Extra half-gallon for thunderstorm
    fuel_level -= extra_fuel_consumed
    print(f"Heavy rain and wind from the thunderstorm consumed an extra {extra_fuel_consumed} gallon of fuel.")

    # Consequence 2: Chance of alternator failure (due to water/electrical issues)
    if random.random() < 0.20: # 20% chance of alternator failure
        print("A sudden jolt or water splash seems to have affected the alternator...")
        handle_alternator_failure()

def handle_tornado_warning():
    print("\n🌪️ Tornado warning! Seek shelter immediately or try to outrun it.")
    # Implement consequences

def handle_alternator_failure():
    print("\n🚨 Alternator failure! Your battery is draining. Find a repair shop soon!")
    # Implement consequences

def handle_alternator_repair():
    print("\n🔧 Alternator repaired! Good to go.")

def handle_engine_trouble():
    print("\n⚙️ Engine trouble! Your vehicle is overheating. Pull over and let it cool.")
    # Implement consequences

def handle_dead_zone():
    print("\n📵 Entering a dead zone. No GPS or cell service. Rely on your map!")
    # Implement consequences

def handle_badlands_shortcut():
    print("\n🏜️ Taking a shortcut through the Badlands. Rough terrain, but could save time.")
    # Implement consequences

def handle_spoiled_commods():
    global CARGO_CATALOG
    if CARGO_CATALOG["Commods (Govt Cheese)"]["qty"] > 0:
        spoiled_amount = random.randint(1, min(5, CARGO_CATALOG["Commods (Govt Cheese)"]["qty"]))
        CARGO_CATALOG["Commods (Govt Cheese)"]["qty"] -= spoiled_amount
        print(f"\n⚠️ ALERT: {spoiled_amount} units of Commods (Govt Cheese) have spoiled due to improper storage!")
        if CARGO_CATALOG["Commods (Govt Cheese)"]["qty"] == 0:
            print("All your Commods (Govt Cheese) are now gone.")

def play_game():
    global wallet, CARGO_CATALOG, fuel_level, FUEL_CONSUMPTION_PER_DAY
    intro()
    rapid_city_market()

    days_on_road = 0
    while True: # Main game loop, assuming destination or failure condition will break this
        days_on_road += 1
        print(f"\n--- DAY {days_on_road} ---")

        # Fuel consumption
        fuel_level -= FUEL_CONSUMPTION_PER_DAY
        print(f"Consumed {FUEL_CONSUMPTION_PER_DAY} gallons of fuel. Remaining fuel: {fuel_level:.1f} gallons.")

        if fuel_level <= 0:
            print("\n⛽ You have run out of fuel! Game Over.")
            break # Exit game loop

        # Random event each day
        event_chance = random.random()
        if event_chance < 0.10:
            handle_snowstorm()
        elif event_chance < 0.20:
            handle_thunderstorm()
        elif event_chance < 0.25:
            handle_tornado_warning()
        elif event_chance < 0.35:
            handle_alternator_failure()
        elif event_chance < 0.40:
            handle_engine_trouble()
        elif event_chance < 0.45:
            handle_dead_zone()
        elif event_chance < 0.50:
            handle_badlands_shortcut()
        elif event_chance < 0.60: # New event for spoiled commods
            handle_spoiled_commods()
        else:
            print("It was an uneventful day on the road.")

        # Display current status in the main loop as well
        print("\n=======================================================")
        used_space = calculate_current_trunk_space()
        print(f"💰 WALLET: ${wallet:.2f}  |  📦 TRUNK SPACE: {used_space} / {TRUNK_CAPACITY} Slots | ⛽ FUEL: {fuel_level:.1f} gal | 🧀 CHEESE: {CARGO_CATALOG['Commods (Govt Cheese)']['qty']}")
        print("=======================================================")
        print("Item Name            | Qty   | Grid Space | Max/Stack ")
        print("-------------------------------------------------------")
        for item, details in CARGO_CATALOG.items():
            if details["qty"] > 0:
                print(f"{item:<20} | {details['qty']:<5} | {details['space']:<10} | Up to {details['max_stack']}")
        print("-------------------------------------------------------")


        # Simulate player action or decision
        action = input("\nWhat do you want to do? (e.g., 'continue', 'check inventory', 'repair', 'rest', 'quit'): ").lower().strip()
        if action == 'continue':
            print("Continuing your journey...")
            # This is where you'd add conditions to reach a destination or run out of resources
        elif action == 'check inventory':
            # Inventory is already displayed, but could add more detail here if needed
            pass
        elif action == 'quit':
            print("Exiting game.")
            break
        else:
            print("Invalid action. Try again.")
```

This test will call the `market_price_fluctuation` function and display the prices of items before and after the fluctuation. Since price changes are random, you may see different results each time you run this test.

```python
print("--- Testing Market Price Fluctuation ---")

# Make a deep copy of CARGO_CATALOG to preserve initial state for comparison
# We need to import copy module for deepcopy
import copy
initial_cargo_catalog_state = copy.deepcopy(CARGO_CATALOG)

print("\nInitial Prices of Fluctuating Items:")
items_to_fluctuate = [item for item in CARGO_CATALOG if item not in ["Gravel Road Map", "Harmonica", "First Aid Kit", "Mechanic Toolbox", "Pellet Rifle Case"]]
for item in items_to_fluctuate:
    print(f"- {item:<20}: ${initial_cargo_catalog_state[item]['cost']:.2f}")

# Call the market price fluctuation function
market_price_fluctuation()

print("\nPrices After Fluctuation:")
prices_changed = False
for item in items_to_fluctuate:
    current_price = CARGO_CATALOG[item]['cost']
    initial_price = initial_cargo_catalog_state[item]['cost']
    if current_price != initial_price:
        change_percent = ((current_price - initial_price) / initial_price) * 100
        print(f"- {item:<20}: ${current_price:.2f} (Changed from ${initial_price:.2f}, {'Increased' if change_percent > 0 else 'Decreased'} by {abs(change_percent):.2f}%) ")
        prices_changed = True
    else:
        print(f"- {item:<20}: ${current_price:.2f} (No change)")

if not prices_changed:
    print("\nNote: No prices changed in this run. Fluctuation is random, so this can happen.")
else:
    print("\nPrices have been updated in the global CARGO_CATALOG.")

print("--- End Market Price Fluctuation Test ---")
```
