### Lakota Alphabet and Pronunciation

The Lakota language uses a modified Latin alphabet. Here are some key points about its pronunciation:

*   **Vowels:** Lakota has oral vowels (a, i, u, e, o) and nasalized vowels (aŋ, iŋ, uŋ, eŋ, oŋ). Nasalized vowels are indicated by a hook (ogonek) or 'ŋ' after the vowel, and they are pronounced by letting air pass through the nose as well as the mouth.
*   **Consonants:** Many consonants are similar to English, but some have distinct pronunciations:
    *   **Aspirated consonants (ph, th, ch, kh, pȟ, tȟ, čh, kȟ):** These are pronounced with a puff of air, similar to the 'p' in 'pot' or 't' in 'top'.
    *   **Ejective consonants (p', t', k', č'):** These are pronounced by building up air pressure behind the consonant and then releasing it abruptly, creating a popping sound.
    *   **Glottal Stop (ʼ):** This is a brief pause or break in the airflow, like the sound between the two parts of 'uh-oh'.
    *   **Fricatives (ȟ, š, ž):**
        *   **ȟ:** A harsh, guttural sound, like the 'ch' in Scottish 'loch'.
        *   **š:** Similar to 'sh' in 'ship'.
        *   **ž:** Similar to 's' in 'pleasure' or 'g' in 'beige'.

Practicing these sounds is essential. You can refer to resources like the `/content/Lakota language.py` file you provided, or listen to native speakers, to get a better grasp of the phonetic nuances.

### Interactive Vocabulary Quiz

Here's a simple function that creates a quiz from your `vocabulary_df`. It will randomly select words and ask you to translate them.

````python
import random

def run_vocabulary_quiz(df, num_questions=5):
    """
    Runs a simple interactive vocabulary quiz.

    Args:
        df (pd.DataFrame): DataFrame with 'Lakota' and 'English' columns.
        num_questions (int): Number of questions for the quiz.
    """
    if df.empty:
        print("Vocabulary list is empty. Cannot run quiz.")
        return

    print(f"\n--- Starting Lakota Vocabulary Quiz ({num_questions} questions) ---")
    correct_answers = 0
    questions = df.sample(n=min(num_questions, len(df)), replace=False).to_dict(orient='records')

    for i, q in enumerate(questions):
        lakota_word = q['Lakota']
        english_translation = q['English']

        prompt_type = random.choice(['lakota_to_english', 'english_to_lakota'])

        if prompt_type == 'lakota_to_english':
            user_answer = input(f"\nQuestion {i+1}: What does '{lakota_word}' mean in English? ").strip()
            correct_ans = english_translation
            lang_pair = "English"
        else: # english_to_lakota
            user_answer = input(f"\nQuestion {i+1}: What is the Lakota word for '{english_translation}'? ").strip()
            correct_ans = lakota_word
            lang_pair = "Lakota"

        if user_answer.lower() == correct_ans.lower():
            print("Correct!")
            correct_answers += 1
        else:
            print(f"Incorrect. The correct {lang_pair} translation was: '{correct_ans}'")

    print(f"\n--- Quiz Over ---")
    print(f"You got {correct_answers} out of {len(questions)} questions correct.")

# Run the quiz with the existing vocabulary_df
# You can change num_questions to practice more words
run_vocabulary_quiz(vocabulary_df, num_questions=3)
````

### Lakota Linguistic Engine (Buechel-Manhart Morphology & Syntax)

This `LakotaLinguisticEngine` class is designed based on the structural rules described in "A Dictionary-Oie Wowapi Wan of Teton Sioux" (1983). It encompasses several advanced linguistic features:

*   **Word Order:** Transforms English SVO (Subject-Verb-Object) to Lakota SOV (Subject-Object-Verb).
*   **Polysynthesis:** Integrates pronouns directly into verb roots as prefixes/infixes (e.g., 'I see' becomes 'wawaŋyáŋka').
*   **Vowel Ablaut:** Predictably mutates terminal root vowels based on syntax, specifically for future tense.
*   **Gendered Speech:** Appends explicit trailing clitics ('yeló' for male, 'kštó' for female) based on the speaker's gender.
*   **Morphological Rules:** Includes methods for diminutives, nominalization, contraction, reduplication, and modifier degrees.

This engine will be highly valuable for understanding and generating grammatically structured Lakota phrases.

````python
import re

class LakotaLinguisticEngine:
    def __init__(self):
        # ----------------------------------------------------------------------
        # 1. Lexicon & Vocabulary Data Subsystem
        # ----------------------------------------------------------------------
        self.vocabulary = {
            "bear": "mathó",
            "man": "wičháša",
            "see": "waŋyáŋka",
            "love": "theȟíla",
            "wise": "kšápa",
            "good": "wašté",
            "water": "mní",
            "drink": "yatkáŋ",
            "woman": "wíŋyaŋ",
            "eat": "yúta",
            "speak": "iyápi",
            "go": "nážiŋ",
            "come": "hí",
            "sing": "lówaŋ",
            "run": "íyaŋka"
        }

        # Pronominal prefix systems (First Class Active)
        self.active_prefixes = {
            "I": "wa",
            "you": "ya"
        }

        # Speaker Gender Declarative Enclitics (Page 26, 31)
        self.gender_enclitics = {
            "male": "yeló",
            "female": "kštó"
        }

        # ----------------------------------------------------------------------
        # 2. Rule 10: Contraction Mapping Table (Page 24)
        # ----------------------------------------------------------------------
        self.contraction_rules = {
            "ata": "al",
            "eća": "el",
            "ća": "l",
            "ǧa": "ȟ",
            "hča": "ȟ",
            "ja": "ś",
            "ka": "g",
            "pa": "b",
            "tu": "l",
            "uta": "ul",
            "yaŋ": "ye",
            "za": "s"
        }

        # ----------------------------------------------------------------------
        # 3. Explicit Reduplication Matrix (Page 24)
        # ----------------------------------------------------------------------
        self.reduplication_matrix = {
            "nakeća": "nagnakeća",
            "nića": "nignića",
            "leǧa": "leǧleǧa",
            "leža": "leśleža",
            "čeća": "čekčeća",
            "lila": "liglila",
            "luta": "lugluta",
            "zica": "zigsica"
        }

        # ----------------------------------------------------------------------
        # 4. Irregular Verb Conjugations (Example for demonstration)
        # This would ideally be much more extensive and perhaps loaded from a file.
        # Format: {English_verb_root: {tense: Lakota_conjugated_form}}
        # ----------------------------------------------------------------------
        self.irregular_verb_conjugations = {
            "be": {
                "present": "waŋyáŋka", # Simplified for example, actual Lakota 'to be' is complex
                "future": "waŋyáŋkapi kta"
            },
            "have": {
                "present": "óhiya",
                "future": "óhiya kta"
            }
        }

    # --------------------------------------------------------------------------
    # Morphological Processing Methods
    # --------------------------------------------------------------------------
    def apply_diminutive(self, noun_root: str) -> str:
        """Applies Rule B-3: Appends the diminutive marker '-la' (Page 23)"""
        return f"{noun_root}la"

    def apply_nominalization(self, verb_root: str) -> str:
        """Applies Rule B-5: Appends the animate/collective noun marker '-pi' (Page 23)"""
        return f"{verb_root}pi"

    def apply_contraction(self, word: str) -> str:
        """Applies Rule 10: Truncates standard root endings to contracted forms (Page 24)"""
        for ending, contraction in self.contraction_rules.items():
            if word.endswith(ending):
                base = word[:-len(ending)]
                return f"{base}{contraction}"
        return word

    def apply_reduplication(self, word: str) -> str:
        """Applies Rule B-1: Generates the intensive/plural reduplicated form (Page 24)"""
        clean_word = word.lower().strip()
        if clean_word in self.reduplication_matrix:
            return self.reduplication_matrix[clean_word]

        # Algorithmic fallback rule for basic CV-CVC roots
        if len(clean_word) <= 5:
            return f"{clean_word[:3]}{clean_word}"
        return f"[Manual Validation Required for: {word}]"

    def generate_modifier_degree(self, adjective: str, degree: str) -> str:
        """Applies Degree Modifiers: Comparative 'sam' and Superlative 'íyotaŋ' (Page 26)"""
        adj_root = self.vocabulary.get(adjective.lower(), adjective)
        if degree == "comparative":
            return f"sam {adj_root}"
        elif degree == "superlative":
            return f"íyotaŋ {adj_root}"
        return adj_root

    def conjugate_irregular_verb(self, verb_root: str, tense: str = "present") -> str or None:
        """Handles irregular verb conjugations.

        Args:
            verb_root (str): The English root of the verb (e.g., 'be', 'have').
            tense (str): The desired tense ('present', 'future', etc.).

        Returns:
            str or None: The Lakota conjugated form if irregular and found, else None.
        """
        normalized_verb = verb_root.lower()
        if normalized_verb in self.irregular_verb_conjugations:
            return self.irregular_verb_conjugations[normalized_verb].get(tense.lower())
        return None

    # --------------------------------------------------------------------------
    # Syntactical Transformation Pipeline
    # --------------------------------------------------------------------------
    def process_sentence(self, subject: str, obj: str, verb: str, speaker_gender: str = "male", is_future: bool = False) -> dict:
        """
        Maps English SVO ideas into Lakota SOV sequences, processing pronoun
        synthesis, gendered speech endings, and future tense vowel ablaut.
        """
        english_svo_layout = f"{subject} + {verb} + {obj}"

        # 1. Base Root Extractions
        # First, check for irregular conjugation
        lakota_verb_root = self.conjugate_irregular_verb(verb, tense="future" if is_future else "present")

        if lakota_verb_root is None: # If not irregular, use regular vocabulary lookup
            lakota_verb_root = self.vocabulary.get(verb.lower(), verb)

        lakota_obj_root = self.vocabulary.get(obj.lower(), obj)

        # 2. Polysynthetic Pronoun Synthesis
        if subject in self.active_prefixes:
            prefix = self.active_prefixes[subject]
            # Only apply prefix if the irregular verb isn't already fully conjugated with it
            if not lakota_verb_root.startswith(prefix):
                 synthesized_verb = f"{prefix}{lakota_verb_root}"
            else:
                synthesized_verb = lakota_verb_root
            lakota_subject_display = f"[{subject} fused inside verb]"
        else:
            synthesized_verb = lakota_verb_root
            lakota_subject_display = self.vocabulary.get(subject.lower(), subject)

        # 3. Ablaut Subsystem (Vowel Shifting for Tense - Page 28, 31)
        # Note: Ablaut might not apply to already irregular/fully conjugated verbs
        # This logic assumes ablaut is primarily for regular verbs.
        if is_future and self.conjugate_irregular_verb(verb, tense="future" if is_future else "present") is None: # Only apply ablaut if not an irregular verb already handled for tense
            # Rule 2: terminal a/aŋ shifts to i/iŋ before future particle 'kta/kte'
            if synthesized_verb.endswith("a"):
                synthesized_verb = synthesized_verb[:-1] + "i"
            elif synthesized_verb.endswith("aŋ"):
                synthesized_verb = synthesized_verb[:-2] + "iŋ"
            tense_marker = "kta"
        else:
            tense_marker = ""

        # 4. Enclitics & Terminal Particle Selection
        enclitic = self.gender_enclitics.get(speaker_gender.lower(), "")

        # 5. Pipeline Assembly
        if subject in self.active_prefixes:
            sequence = [lakota_obj_root, synthesized_verb, tense_marker, enclitic]
        else:
            sequence = [lakota_subject_display, lakota_obj_root, synthesized_verb, tense_marker, enclitic]

        lakota_sov_output = " ".join(filter(None, sequence))

        return {
            "english_svo": english_svo_layout,
            "lakota_sov": lakota_sov_output,
            "matrix_breakdown": {
                "Subject Segment": lakota_subject_display,
                "Object Segment": lakota_obj_root,
                "Verb Root Stem": verb, # Keep original English root here for clarity
                "Synthesized Predicate": synthesized_verb,
                "Tense Trigger": tense_marker if is_future else "Present / Unmarked",
                "Gender Enclitic": enclitic
            }
        }
````

                return word[:-len(ending)] + contraction
        return word

    def apply_reduplication(self, word: str) -> str:
        """Applies Rule B-1: Generates intensive/plural forms (Page 24)"""
        clean = word.lower().strip()
        if clean in self.reduplication_matrix:
            return self.reduplication_matrix[clean]
        if len(clean) <= 5:
            return clean[:3] + clean
        return f"[Manual Validation Needed: {word}]"

    def process_sentence(self, subject: str, obj: str, verb: str, speaker_gender: str = "male", is_future: bool = False) -> dict:
        """Maps English SVO structure into Lakota SOV syntax with morphophonemic rules."""
        lakota_obj = self.vocabulary.get(obj.lower(), obj)
        lakota_verb = self.vocabulary.get(verb.lower(), verb)

        if subject in self.active_prefixes:
            synthesized_verb = f"{self.active_prefixes[subject]}{lakota_verb}"
            subject_display = f"[{subject} integrated]"

            synthesized_verb = lakota_verb

        enclitic = self.gender_enclitics.get(speaker_gender.lower(), "")

        if subject in self.active_prefixes:
            sequence = [lakota_obj, synthesized_verb, tense_marker, enclitic]

            sequence = [subject_display, lakota_obj, synthesized_verb, tense_marker, enclitic]

        return {
            "english_svo": f"{subject} + {verb} + {obj}",
            "lakota_sov": " ".join(filter(None, sequence))
        }
