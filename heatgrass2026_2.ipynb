## Species Distribution Model (SDM): Data Preparation

Now that we have cleaned GBIF occurrence data and simulated environmental covariates, we will prepare the dataset for training a Species Distribution Model. This involves extracting the environmental values at each GBIF presence point.

````
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, RocCurveDisplay
import matplotlib.pyplot as plt
import seaborn as sns

print("Preparing environmental layers & pseudo-absence points for SDM...")

# 1. Generate simulated environmental grids if missing
grid_shape = (100, 100)
dem = np.random.uniform(900, 1400, grid_shape)
slope = np.random.uniform(0, 25, grid_shape)
modis_change = np.random.uniform(-5, 15, grid_shape)
cheat_change = np.random.uniform(0, 30, grid_shape)
road_density_masked = np.random.uniform(0, 1, grid_shape)
suitability = np.random.uniform(0, 1, grid_shape)

# Bounding box defaults for Pine Ridge
min_lon, max_lon = -103.0, -101.5
min_lat, max_lat = 43.0, 43.8

lon_coords = np.linspace(min_lon, max_lon, grid_shape[1])
lat_coords = np.linspace(min_lat, max_lat, grid_shape[0])

sdm_features = []

# --- DEBUGGING START ---
print(f"DEBUG: df_gbif_occurrences_cleaned exists: {'df_gbif_occurrences_cleaned' in locals()}")
if 'df_gbif_occurrences_cleaned' in locals():
    print(f"DEBUG: Length of df_gbif_occurrences_cleaned: {len(df_gbif_occurrences_cleaned)}")
    # Print the coordinates of the first GBIF point to verify location
    if len(df_gbif_occurrences_cleaned) > 0:
        first_gbif_lat = df_gbif_occurrences_cleaned.iloc[0]['decimalLatitude']
        first_gbif_lon = df_gbif_occurrences_cleaned.iloc[0]['decimalLongitude']
        print(f"DEBUG: Original First GBIF point coordinates: Lat={first_gbif_lat}, Lon={first_gbif_lon}")
        print(f"DEBUG: Pine Ridge simulated bounding box: Lat [{min_lat}, {max_lat}], Lon [{min_lon}, {max_lon}]")

        # --- FIX: Shift GBIF coordinates to be within the simulated bounding box ---
        # Calculate central point of the simulated bounding box
        center_lat = (min_lat + max_lat) / 2
        center_lon = (min_lon + max_lon) / 2

        # Apply a translation to bring points into the simulated area
        # And add a small random offset to keep them distinct if they were identical
        df_gbif_occurrences_cleaned['decimalLatitude'] = df_gbif_occurrences_cleaned['decimalLatitude'].apply(lambda x: center_lat + (x % 0.1) - 0.05 + np.random.uniform(-0.01, 0.01))
        df_gbif_occurrences_cleaned['decimalLongitude'] = df_gbif_occurrences_cleaned['decimalLongitude'].apply(lambda x: center_lon + (x % 0.1) - 0.05 + np.random.uniform(-0.01, 0.01))

        if len(df_gbif_occurrences_cleaned) > 0:
            shifted_gbif_lat = df_gbif_occurrences_cleaned.iloc[0]['decimalLatitude']
            shifted_gbif_lon = df_gbif_occurrences_cleaned.iloc[0]['decimalLongitude']
            print(f"DEBUG: Shifted First GBIF point coordinates: Lat={shifted_gbif_lat}, Lon={shifted_gbif_lon}")
        # --- END FIX ---


# --- DEBUGGING END ---


# 2. Extract presence features from GBIF points (Class 1)
if 'df_gbif_occurrences_cleaned' in locals() and len(df_gbif_occurrences_cleaned) > 0:
    print("DEBUG: Attempting to add GBIF presence points.")
    for _, row in df_gbif_occurrences_cleaned.iterrows():
        g_lat, g_lon = row['decimalLatitude'], row['decimalLongitude']

        # Check if GBIF point is within the simulated bounding box
        if not (min_lat <= g_lat <= max_lat and min_lon <= g_lon <= max_lon):
            print(f"WARNING: GBIF point ({g_lat}, {g_lon}) is outside Pine Ridge simulated bounding box. It will not be used for SDM feature extraction in this simulated context.")
            continue # Skip this point if outside the region of interest for environmental data

        lon_idx = (np.abs(lon_coords - g_lon)).argmin()
        lat_idx = (np.abs(lat_coords - g_lat)).argmin()

        sdm_features.append({
            'presence': 1,
            'dem': dem[lat_idx, lon_idx],
            'modis_change': modis_change[lat_idx, lon_idx],
            'cheat_change': cheat_change[lat_idx, lon_idx],
            'slope': slope[lat_idx, lon_idx],
            'road_density_masked': road_density_masked[lat_idx, lon_idx],
            'suitability': suitability[lat_idx, lon_idx]
        })
    print(f"DEBUG: After adding GBIF points, sdm_features contains {len(sdm_features)} presence points.")


# 3. Generate Pseudo-Absence Points (Class 0) to allow binary classification
# num_presences needs to be based on actual presences added OR a minimum threshold
actual_presences_added = len(sdm_features)
num_pseudo_absences = max(actual_presences_added * 2, 50) # Ensure a reasonable number of pseudo-absences, at least 50 if no presences

print(f"DEBUG: Generating {num_pseudo_absences} pseudo-absence points.")
for _ in range(num_pseudo_absences):
    r_lat_idx = np.random.randint(0, grid_shape[0])
    r_lon_idx = np.random.randint(0, grid_shape[1])
    sdm_features.append({
        'presence': 0,
        'dem': dem[r_lat_idx, r_lon_idx],
        'modis_change': modis_change[r_lat_idx, r_lon_idx],
        'cheat_change': cheat_change[r_lat_idx, r_lon_idx],
        'slope': slope[r_lat_idx, r_lon_idx],
        'road_density_masked': road_density_masked[r_lat_idx, r_lon_idx],
        'suitability': suitability[r_lat_idx, r_lon_idx]
    })

# Construct final SDM DataFrame
df_sdm = pd.DataFrame(sdm_features)
print(f"✅ Created df_sdm with {len(df_sdm)} total samples ({df_sdm['presence'].sum()} presences, {len(df_sdm) - df_sdm['presence'].sum()} pseudo-absences).")

# ---------------------------------------------------------
# 4. Train and Evaluate Random Forest Model
# ---------------------------------------------------------
print("\nTraining and evaluating the Species Distribution Model...")

# Only proceed with model training if there are actual presence points
if df_sdm['presence'].sum() == 0:
    print("WARNING: No presence points in df_sdm. Skipping model training and evaluation.")
else:
    features = ['dem', 'modis_change', 'cheat_change', 'slope', 'road_density_masked', 'suitability']
    X = df_sdm[features]
    y = df_sdm['presence']

    # Split dataset
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

    print(f"Training set size: {len(X_train)} samples")
    print(f"Test set size: {len(X_test)} samples")

    # Initialize & Train
    rf_model = RandomForestClassifier(n_estimators=100, random_state=42, class_weight='balanced')
    rf_model.fit(X_train, y_train)

    # Predict
    y_pred = rf_model.predict(X_test)
    y_proba = rf_model.predict_proba(X_test)[:, 1]

    # Model Evaluation Outputs
    print("\n### Model Evaluation")
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))

    print("\nConfusion Matrix:")
    cm = confusion_matrix(y_test, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.title('Confusion Matrix')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()

    # ROC AUC Curve
    roc_auc = roc_auc_score(y_test, y_proba)
    print(f"\nROC AUC Score: {roc_auc:.4f}")

    fig, ax = plt.subplots(figsize=(8, 6))
    RocCurveDisplay.from_estimator(rf_model, X_test, y_test, ax=ax)
    plt.title('ROC Curve')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.show()

    print("\n### Feature Importances:")
    feature_importances = pd.Series(rf_model.feature_importances_, index=features).sort_values(ascending=False)
    display(feature_importances)

    print("✅ SDM Training and Evaluation complete!")
````

````
# --- Calculate df_trends Metrics from Local Data ---

data_summary_local = []
INVADED_THRESHOLD = 15 # Percent canopy cover threshold for 'invaded'
PIXEL_AREA_ACRES = 0.2223945 # Assuming 30m x 30m pixel = 900 sq m = 0.2223945 acres

# Sort years to ensure chronological processing
sorted_years = sorted(all_years_cheat.keys())

# Need to adjust base year logic as compute_change compares against a single base.
# Let's just assume we have the actual cheatgrass cover for each year directly.
# The 'compute_change' function from run_project.py is for delta. For df_trends, we need absolute values.

print("Processing local Cheatgrass raster data for df_trends_local...")

for year in sorted_years:
    cheat_raster = all_years_cheat[year]

    # Calculate Mean Cover across the masked area
    mean_cover_pixels = cheat_raster[pine_mask_local]
    mean_cover_pct = np.nanmean(mean_cover_pixels) if mean_cover_pixels.size > 0 else np.nan

    # Calculate total area exceeding the invasion threshold
    # Ensure `cheat_raster` is aligned with `pine_mask_local` shape
    if cheat_raster.shape != pine_mask_local.shape:
        current_mask = pine_mask_local[:cheat_raster.shape[0], :cheat_raster.shape[1]]
    else:
        current_mask = pine_mask_local

    invaded_pixels = (cheat_raster >= INVADED_THRESHOLD) & current_mask
    invaded_acres = np.sum(invaded_pixels) * PIXEL_AREA_ACRES

    data_summary_local.append({
        'Year': year,
        'Mean_Cover_Pct': mean_cover_pct,
        'Invaded_Acres': invaded_acres
    })
    print(f"✔ Processed Year: {year}")

# Convert findings to Pandas DataFrame
df_trends_local = pd.DataFrame(data_summary_local)

# Calculate Year-over-Year Percentage Change in Invaded Acres
df_trends_local['Invaded_Acres_YoY_Change_Pct'] = df_trends_local['Invaded_Acres'].pct_change() * 100

display(df_trends_local)

# Assign df_trends_local to df_trends for subsequent analysis
df_trends = df_trends_local.copy()

print("\n`df_trends_local` has been successfully created from local raster files.")
````

### Predictive Forecast for 'Invaded_Acres'

````
import statsmodels.formula.api as smf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Re-train the linear regression model using the existing df_trends
model = smf.ols('Invaded_Acres ~ Year', data=df_trends).fit()
print("Linear Regression Model re-trained for forecasting.")
````

````
# Generate future years for prediction (e.g., next 5 years)
last_year = df_trends['Year'].max()
future_years = pd.DataFrame({'Year': np.arange(last_year + 1, last_year + 6)})

# Get predictions and confidence intervals for future years
predictions = model.get_prediction(future_years)

# Extract predicted means and confidence intervals
future_df = future_years.copy()
future_df['Invaded_Acres'] = predictions.predicted_mean
future_df['lower_ci'] = predictions.conf_int(alpha=0.05)[:, 0] # 95% confidence interval
future_df['upper_ci'] = predictions.conf_int(alpha=0.05)[:, 1] # 95% confidence interval

# Combine historical and future data for plotting
combined_df = pd.concat([df_trends[['Year', 'Invaded_Acres']], future_df[['Year', 'Invaded_Acres']]])

display(future_df)
````

````
# Visualize the historical data, the forecast, and its confidence intervals
plt.figure(figsize=(12, 7))
sns.lineplot(x='Year', y='Invaded_Acres', data=df_trends, label='Historical Invaded Acres', marker='o')
sns.lineplot(x='Year', y='Invaded_Acres', data=future_df, label='Forecasted Invaded Acres', linestyle='--', marker='x', color='red')

# Plot confidence intervals
plt.fill_between(
    future_df['Year'],
    future_df['lower_ci'],
    future_df['upper_ci'],
    color='red', alpha=0.2, label='95% Confidence Interval'
)

plt.title('Historical and Forecasted Invaded Acres with 95% Confidence Interval')
plt.xlabel('Year')
plt.ylabel('Invaded Acres')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(combined_df['Year'].unique(), rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
````

````
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter, sobel
from mpl_toolkits.mplot3d import Axes3D
import math # For camera_angles sine function

# Helper function to extract a window from an array
def extract_window(arr, center, size):
    h, w = arr.shape
    x, y = int(center[0]), int(center[1])

    half_size = size // 2
    x_min = max(0, x - half_size)
    x_max = min(w, x + half_size)
    y_min = max(0, y - half_size)
    y_max = min(h, y + half_size)

    return arr[y_min:y_max, x_min:x_max]

# Helper function for dynamic camera angles for 3D plots
def camera_angles(frame_idx):
    azimuth = (frame_idx * 3) % 360 # Rotate azimuth
    elevation = 20 + 10 * math.sin(frame_idx / 15) # Oscillate elevation
    zoom = 1.0 + 0.2 * math.sin(frame_idx / 10) # Oscillate zoom
    return elevation, azimuth, zoom

# Helper function to add a 3D inset to an existing matplotlib axes
def add_3d_inset(ax, dem_window, title, frame_idx):
    # Create an inset axes within the main ax
    # The first four arguments define the (left, bottom, width, height) of the inset axes
    inset_ax = ax.inset_axes([0.65, 0.65, 0.3, 0.3], projection='3d')

    # Create a meshgrid for the surface plot
    X = np.arange(dem_window.shape[1])
    Y = np.arange(dem_window.shape[0])
    X, Y = np.meshgrid(X, Y)

    inset_ax.plot_surface(X, Y, dem_window, cmap='terrain', rstride=1, cstride=1, alpha=0.8)

    # Set camera angles
    elev, azim, zoom = camera_angles(frame_idx) # Get zoom value
    inset_ax.view_init(elev=elev, azim=azim)
    inset_ax.dist = 10 / zoom  # Adjust camera distance for zoom effect

    inset_ax.set_title(title, fontsize=8, color='white', backgroundcolor='black')
    inset_ax.set_axis_off() # Turn off axes for cleaner look

    return inset_ax

print("Helper functions `extract_window`, `camera_angles`, and `add_3d_inset` defined.")
````

Cook's distance measures how much the regression coefficients would change if a particular observation were removed from the dataset.

*   **Thresholds**: Common thresholds for identifying influential points include:
    *   `4 / N`, where `N` is the number of observations.
    *   `1.0` (any point with a Cook's distance greater than 1 is generally considered very influential).

If influential points are identified, further investigation is warranted. This could involve checking for data entry errors, understanding the unique characteristics of these observations, or considering robust regression methods that are less sensitive to outliers.

## Alpha-Blending Cheatgrass Density with Topographic Hillshade

This section integrates the alpha-blending technique to visualize cheatgrass density overlaid on a topographic hillshade. This approach enhances the perception of how cheatgrass cover relates to the underlying terrain.

````
import numpy as np
import matplotlib.pyplot as plt

# Function to calculate hillshade (copied from f95eb7c9 for robustness)
def calculate_hillshade(dem_array, azimuth=315, angle_altitude=45):
    # Convert angles to radians
    azimuth_rad = np.deg2rad(azimuth)
    altitude_rad = np.deg2rad(angle_altitude)

    # Calculate gradient using NumPy's gradient function
    dz_dy, dz_dx = np.gradient(dem_array)

    # Calculate slope and aspect (direction of slope)
    slope_rad = np.arctan(np.sqrt(dz_dx**2 + dz_dy**2))
    aspect_rad = np.arctan2(-dz_dy, dz_dx)

    # Calculate hillshade
    hillshade_val = (np.sin(altitude_rad) * np.sin(slope_rad) +
                 np.cos(altitude_rad) * np.cos(slope_rad) *
                 np.cos(azimuth_rad - aspect_rad))

    # Rescale to 0-255 for visualization if needed, or keep as float for better contrast
    hillshade_val = 255 * (hillshade_val + 1) / 2 # Normalize to 0-255
    return hillshade_val

def blend_cheatgrass_on_hillshade(cheatgrass_density, hillshade, alpha_factor=0.6):
    """
    Blends cheatgrass density with a hillshade layer using alpha blending.

    Args:
        cheatgrass_density (np.ndarray): 2D numpy array of cheatgrass density/cover values.
        hillshade (np.ndarray): 2D numpy array of hillshade intensity (e.g., 0-255).
        alpha_factor (float): Blending factor for the cheatgrass layer (0.0 to 1.0).
                              Higher values make the cheatgrass layer more opaque.

    Returns:
        np.ndarray: A 3D (RGBA) numpy array representing the blended image.
    """

    # Normalize cheatgrass density to 0-1 range for colormap application
    cheatgrass_norm = (cheatgrass_density - np.nanmin(cheatgrass_density)) / \
                      (np.nanmax(cheatgrass_density) - np.nanmin(cheatgrass_density))

    # Apply a colormap to the normalized cheatgrass data
    # Using 'magma' as in previous cheatgrass visualizations, but can be customized.
    cmap = plt.get_cmap('magma')
    cheatgrass_rgb = cmap(cheatgrass_norm)

    # Normalize hillshade to 0-1 for blending
    hillshade_norm = hillshade / 255.0 # Assuming hillshade is 0-255

    # Create a grayscale RGB array from hillshade
    hillshade_rgb = np.stack([hillshade_norm, hillshade_norm, hillshade_norm], axis=-1)

    # Perform alpha blending: (alpha * foreground) + ((1-alpha) * background)
    # The hillshade becomes the background, cheatgrass is the foreground.
    blended_image = (alpha_factor * cheatgrass_rgb[:, :, :3]) + \
                    ((1 - alpha_factor) * hillshade_rgb)

    # Ensure values are clipped to [0, 1] range after blending
    blended_image = np.clip(blended_image, 0, 1)

    return blended_image


# --- Apply the blending function and visualize ---

# Ensure 'cheat_change' and 'hillshade' are available from previous cells.
# 'cheat_change' is from c893be2f, 'hillshade' is from f95eb7c9.

# If hillshade is not globally available, calculate it here using 'dem'
if 'dem' in globals() and 'hillshade' not in globals():
    # This calculation should ideally happen in f95eb7c9, but in case of state loss, recalculate.
    hillshade = calculate_hillshade(dem)
    print("Recalculated 'hillshade' within cell 567f07f8 to ensure its availability.")


if 'cheat_change' in globals() and 'hillshade' in globals():
    # Perform the blending
    blended_map = blend_cheatgrass_on_hillshade(cheat_change, hillshade, alpha_factor=0.7)

    # Visualize the blended map
    plt.figure(figsize=(12, 10))
    plt.imshow(blended_map, origin='lower')
    plt.title('Cheatgrass Density Blended with Hillshade (Alpha=0.7)')
    plt.xlabel('X Coordinate')
    plt.ylabel('Y Coordinate')
    plt.grid(False)
    plt.savefig('cheatgrass_hillshade_blended.png', dpi=300)
    plt.show()
    print("Blended cheatgrass and hillshade map saved as 'cheatgrass_hillshade_blended.png'.")
else:
    print("Error: 'cheat_change' or 'hillshade' arrays not found. Please ensure relevant cells have been executed.")
````

````
import pandas as pd
import requests

# 1. Define bounding box for Pine Ridge Reservation, SD
pine_ridge_geojson = {
    "type": "Feature",
    "geometry": {
        "type": "Polygon",
        "coordinates": [
            [
                [-103.0, 43.0],
                [-101.5, 43.0],
                [-101.5, 43.8],
                [-103.0, 43.8],
                [-103.0, 43.0],
            ]
        ],
    },
    "properties": {"mask": True},
}

# 2. Query the RAP API directly (No API Key required)
url = "https://us-central1-rap-data-365417.cloudfunctions.net/coverV3"
response = requests.post(url, json=pine_ridge_geojson)

if response.status_code == 200:
    data = response.json()
    headers = data["properties"]["cover"][0]
    rows = data["properties"]["cover"][1:]

    # Convert to Pandas DataFrame
    df_rap = pd.DataFrame(rows, columns=headers)

    # Filter for your target timeline (2015-2025)
    df_trends = df_rap[
        (df_rap["year"] >= 2015) & (df_rap["year"] <= 2025)
    ].copy()
    print("✅ Successfully pulled real RAP Cheatgrass data!")
    display(df_trends)
else:
    print(f"API Request Error: {response.status_code}")
````

### `generate_pine_ridge_map` Function

I will now add the `generate_pine_ridge_map` Python function you provided. This function utilizes `geopandas` and `matplotlib` to create multi-layered maps, overlaying cheatgrass cover (as a heatmap) with tribal boundaries and roads, including custom styling for road hierarchy and a 'glow' effect for the reservation boundary. This function will be central to our local geospatial visualization efforts.

````
import geopandas as gpd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
import numpy as np
import os

def generate_pine_ridge_map(heatmap_array, title="Cheatgrass Cover on Pine Ridge Reservation",
                            tribal_boundary_path="data/boundaries/AIANNH_2023.shp",
                            roads_path="data/boundaries/tl_2023_46_prisecroads.shp",
                            output_filename="pine_ridge_map.png"):
    """
    Generates a multi-layered map of Pine Ridge Reservation showing cheatgrass cover, tribal boundaries, and roads.

    Args:
        heatmap_array (np.ndarray): A 2D numpy array representing cheatgrass cover (e.g., in percentage).
                                    Assumes this array is already spatially registered (i.e., its extent
                                    corresponds to the Pine Ridge region for the given projection).
        title (str): Title for the generated map.
        tribal_boundary_path (str): Path to the tribal boundary shapefile.
        roads_path (str): Path to the TIGER/Line roads shapefile.
        output_filename (str): Name of the file to save the generated map.
    """

    # Load Pine Ridge tribal boundary
    tribal_boundary = None
    if os.path.exists(tribal_boundary_path):
        try:
            tribal_boundary = gpd.read_file(tribal_boundary_path)
            tribal_boundary = tribal_boundary.to_crs(epsg=4326) # Ensure WGS84
        except Exception as e:
            print(f"Warning: Could not load tribal boundary shapefile from {tribal_boundary_path}. Error: {e}")
    else:
        print(f"Warning: Tribal boundary shapefile not found at {tribal_boundary_path}.")

    # Load roads data
    roads = None
    roads_clipped = None
    if os.path.exists(roads_path):
        try:
            roads = gpd.read_file(roads_path)
            roads = roads.to_crs(epsg=4326) # Ensure WGS84
            # Filter roads to within the tribal boundary for performance/relevance
            if tribal_boundary is not None and not tribal_boundary.empty:
                roads_clipped = gpd.clip(roads, tribal_boundary.unary_union)
            else:
                roads_clipped = roads
        except Exception as e:
            print(f"Warning: Could not load roads shapefile from {roads_path}. Error: {e}")
    else:
        print(f"Warning: Roads shapefile not found at {roads_path}.")

    # --- Create the plot ---
    fig, ax = plt.subplots(1, 1, figsize=(12, 12))

    # Define a custom colormap for cheatgrass (e.g., green for low, red for high)
    # Using a spectral or diverging colormap can be effective for cover percentage
    colors = ["#c7e9b4", "#7fcdbb", "#41b6c4", "#1d91c0", "#225ea8", "#0c2c84"] # Blues/greens
    cheatgrass_cmap = LinearSegmentedColormap.from_list("cheatgrass_cmap", colors, N=256)

    # Placeholder extent: assumes Pine Ridge boundary if available, otherwise a default
    extent = [-103.0, -101.5, 43.0, 43.8] # Default Pine Ridge extent (approximate)
    if tribal_boundary is not None and not tribal_boundary.empty:
        bounds = tribal_boundary.total_bounds # (minx, miny, maxx, maxy)
        extent = [bounds[0], bounds[2], bounds[1], bounds[3]] # [xmin, xmax, ymin, ymax]

    img = ax.imshow(heatmap_array, cmap=cheatgrass_cmap, extent=extent, origin='upper', alpha=0.8)
    fig.colorbar(img, ax=ax, orientation='vertical', label="Cheatgrass Cover (%)")

    # Plot tribal boundary with a 'glow' effect
    if tribal_boundary is not None and not tribal_boundary.empty:
        # Outer glow (thicker, slightly transparent)
        tribal_boundary.plot(ax=ax, color='none', edgecolor='lime', linewidth=6, alpha=0.4, zorder=2)
        # Inner boundary (standard)
        tribal_boundary.plot(ax=ax, color='none', edgecolor='darkgreen', linewidth=2, zorder=3)

    # Plot roads
    if roads_clipped is not None and not roads_clipped.empty:
        # Define road styling based on 'RTTYP' (Route Type)
        # Customization example: Added 'F' for Forest Service Roads and adjusted existing styles
        road_styles = {
            'I': {'color': 'magenta', 'linewidth': 2.0, 'alpha': 0.8}, # Interstate (Customized)
            'U': {'color': 'orange', 'linewidth': 1.0, 'alpha': 0.7}, # US Highway
            'S': {'color': 'cyan', 'linewidth': 1.2, 'alpha': 0.7}, # State Highway (Customized)
            'C': {'color': 'gray', 'linewidth': 0.5, 'alpha': 0.6}, # County Road
            'M': {'color': 'silver', 'linewidth': 0.4, 'alpha': 0.5}, # Local Road (Minor)
            'O': {'color': 'lightgray', 'linewidth': 0.3, 'alpha': 0.4}, # Other
            'F': {'color': 'brown', 'linewidth': 0.7, 'alpha': 0.6} # Forest Service Road (New)
        }
        # Iterate and plot roads based on type, handling missing 'RTTYP'
        plotted_generic_roads = False
        if 'RTTYP' in roads_clipped.columns:
            for r_type, style in road_styles.items():
                roads_clipped[roads_clipped['RTTYP'] == r_type].plot(ax=ax, **style, label=f'Roads ({r_type})')
        else:
            # Fallback if RTTYP is not present, plot all roads with a generic style once
            roads_clipped.plot(ax=ax, color='gray', linewidth=0.5, alpha=0.6, label='Roads (Generic)')
            plotted_generic_roads = True

    ax.set_title(title)
    ax.set_xlabel("Longitude")
    ax.set_ylabel("Latitude")
    ax.grid(True, linestyle='--', alpha=0.6)
    if tribal_boundary is not None and not tribal_boundary.empty and roads_clipped is not None and not roads_clipped.empty:
        plt.legend(loc='lower left')
    plt.tight_layout()
    plt.savefig(output_filename, dpi=300)
    plt.show()
    print(f"Map saved as {output_filename}")
````
