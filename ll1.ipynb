```python
class LakotaMorphologyEngine:
    def __init__(self):
        # Suffix change markers from Section 1
        self.ablaut_e_triggers = ["ca", "čhaŋké", "ye", "lo", "yo", "po"]
        self.ablaut_i_triggers = ["na", "naiŋś", "ye", "kta", "kte"]

    def apply_vowel_ablaut(self, verb_stem: str, trailing_particle: str) -> str:
        """
        Executes Rule 1 & 2: Changes terminal 'a' or 'aŋ' to 'e' or 'i'
        depending strictly on the following particle environment context.
        """
        if not (verb_stem.endswith("a") or verb_stem.endswith("aŋ")):
            return verb_stem # Return unmodified if word does not end in a changeable vowel

        # Rule 1: Change a/aŋ to e
        if trailing_particle in self.ablaut_e_triggers:
            if verb_stem.endswith("a"):
                return verb_stem[:-1] + "e"
            elif verb_stem.endswith("aŋ"):
                return verb_stem[:-2] + "eŋ"

        # Rule 2: Change a/aŋ to i before future metrics or conjunctions
        if trailing_particle in self.ablaut_i_triggers:
            if verb_stem.endswith("a"):
                return verb_stem[:-1] + "i"
            elif verb_stem.endswith("aŋ"):
                return verb_stem[:-2] + "iŋ"

        return verb_stem

    def trace_verb_slots(self, absolute_prefix, prep, pronoun, valency, instrument, root_stem) -> str:
        """
        Models Section 'THE FORMATION OF VERBS' structural layout slots.
        Combines semantic bounds cleanly into a single synthetic verb chain.
        """
        slots = [absolute_prefix, prep, pronoun, valency, instrument, root_stem]
        # Clean out empty arguments
        clean_slots = [s for s in slots if s]
        return "".join(clean_slots)
```

```python
def _find_definition_in_dict(data_dict, word, definitions_list):
    """
    Helper function to recursively search for a word as a key within a dictionary
    and append its value (definition) to a list.
    """
    if isinstance(data_dict, dict):
        for key, value in data_dict.items():
            if key == word:
                definitions_list.append(value)
            # Recursively search in nested dictionaries or lists
            _find_definition_in_dict(value, word, definitions_list)
    elif isinstance(data_dict, list):
        for item in data_dict:
            _find_definition_in_dict(item, word, definitions_list)

def lookup_word_definition(word: str) -> list:
    """
    Looks up a word across all loaded linguistic data collections and returns
    a list of its definitions.
    """
    all_definitions = []
    # List of all root-level dictionaries to search
    collections_to_search = [
        adverbs_data,
        adjectives_relationships_data,
        prepositions_data,
        interjections_data
    ]

    for collection_dict in collections_to_search:
        _find_definition_in_dict(collection_dict, word, all_definitions)

    return all_definitions
```

### Automating Simple Lakota Sentence Generation

To automate sentence generation, we'll leverage the `LakotaMorphologyEngine` and the loaded linguistic dictionaries. Given the current available vocabulary (adverbs, prepositions, interjections, and relationship terms, but a limited set of common nouns or diverse verb stems), we'll focus on generating simple sentence structures.

For this demonstration, we'll use a placeholder subject and focus on combining:

1.  A base verb stem (e.g., 'waŋyáŋka' - to see).
2.  A randomly chosen ablaut-triggering particle (from the `LakotaMorphologyEngine`).
3.  One or more randomly chosen adverbs from the `adverbs_data` collection.

```python
import random

def get_random_word(data_dict: dict, category: str = None) -> str:
    """
    Extracts a random word (key) from a given linguistic data dictionary.
    Optionally, from a specific category within the dictionary if it has nested structure.
    """
    if not data_dict:
        return None

    # If a category is specified and exists, try to get words from there
    if category and 'data' in data_dict and category in data_dict['data']:
        words = list(data_dict['data'][category].keys())
        if words:
            return random.choice(words)
    # Otherwise, try to extract words from the top level or 'data' sub-dictionary
    elif 'data' in data_dict:
        all_words_in_data = []
        for sub_cat in data_dict['data']:
            all_words_in_data.extend(list(data_dict['data'][sub_cat].keys()))
        if all_words_in_data:
            return random.choice(all_words_in_data)
    elif 'paradigms' in data_dict:
        # For 'adjectives_relationships_data', we need to dig a bit deeper
        all_words_in_paradigms = []
        for paradigm_type in data_dict['paradigms']:
            if isinstance(data_dict['paradigms'][paradigm_type], dict):
                for key, value in data_dict['paradigms'][paradigm_type].items():
                    if isinstance(value, str): # Simple key-value pair
                        all_words_in_paradigms.append(key)
                    elif isinstance(value, dict): # Nested dictionary (like male_speaker)
                        for sub_key, sub_value in value.items():
                            if isinstance(sub_value, str):
                                all_words_in_paradigms.append(sub_key) # This gets 'vocative', 'possessive'
                            elif isinstance(sub_value, dict): # If it gets even deeper
                                all_words_in_paradigms.extend(list(sub_value.keys()))
            elif isinstance(data_dict['paradigms'][paradigm_type], str):
                 all_words_in_paradigms.append(data_dict['paradigms'][paradigm_type]) # Directly take value if it's a string

        if all_words_in_paradigms:
            # Filter for actual Lakota words rather than descriptive keys like 'root', 'male_speaker'
            lakota_words = [word for word in all_words_in_paradigms if not any(kw in word for kw in ['speaker', 'lineage', 'root', 'prefixes'] ) and len(word) > 1]
            if lakota_words:
                return random.choice(lakota_words)

    elif data_dict: # If it's a flat dictionary
        words = list(data_dict.keys())
        if words:
            return random.choice(words)
    return None

def get_random_relationship_term(adjectives_relationships_data: dict) -> str:
    """
    Extracts a random relationship term from the adjectives_relationships_data.
    """
    relationship_terms = []
    paradigms = adjectives_relationships_data.get('paradigms', {})

    # Extract terms from father_lineage and mother_lineage
    for lineage_type in ['father_lineage', 'mother_lineage']:
        if lineage_type in paradigms:
            lineage_data = paradigms[lineage_type]
            if 'root' in lineage_data:
                relationship_terms.append(lineage_data['root'])

    # Extract terms from sibling perspectives
    for sibling_type in ['siblings_male_perspective', 'siblings_female_perspective']:
        if sibling_type in paradigms and isinstance(paradigms[sibling_type], dict):
            relationship_terms.extend(list(paradigms[sibling_type].values()))

    unique_terms = list(set([term for term in relationship_terms if term]))
    return random.choice(unique_terms) if unique_terms else None

def get_random_possessive_prefix(adjectives_relationships_data: dict) -> str:
    """
    Extracts a random possessive prefix from the adjectives_relationships_data.
    """
    possessive_prefixes = adjectives_relationships_data.get('paradigms', {}).get('possessive_prefixes', {})
    prefixes = list(possessive_prefixes.keys())
    return random.choice(prefixes) if prefixes else None

def generate_simple_lakota_sentence(engine: LakotaMorphologyEngine, verb_stem: str = "waŋyáŋka") -> str:
    """
    Generates a simple Lakota-like sentence using the morphology engine
    and loaded linguistic data, including noun phrases for the subject and object.
    """

    # Helper to generate a subject noun phrase
    def generate_subject_noun_phrase(adjectives_relationships_data_local: dict) -> str:
        # Option 1: Simple generic subject (hardcoded for now as they aren't in JSON directly as nouns)
        if random.random() < 0.3: # 30% chance for a generic subject
            return random.choice(["Wičháša", "Wiŋyáŋ"]) # Man, Woman

        # Option 2: Relationship term, with or without possessive prefix
        relationship_term = get_random_relationship_term(adjectives_relationships_data_local)
        if relationship_term:
            if random.random() < 0.5: # 50% chance to add a possessive prefix
                possessive_prefix = get_random_possessive_prefix(adjectives_relationships_data_local)
                if possessive_prefix:
                    return f"{possessive_prefix} {relationship_term}"
            return relationship_term
        return "Someone" # Fallback if no relationship term found

    # Helper to generate an object noun phrase
    def generate_object_noun_phrase(adjectives_relationships_data_local: dict) -> str:
        # Option 1: Simple generic object (hardcoded for now as they aren't in JSON directly as nouns)
        if random.random() < 0.2: # 20% chance for a generic object
            return random.choice(["Wičháša", "Wiŋyáŋ", "Kola"]) # Man, Woman, Friend

        # Option 2: Relationship term, with or without possessive prefix
        relationship_term = get_random_relationship_term(adjectives_relationships_data_local)
        if relationship_term:
            if random.random() < 0.5: # 50% chance to add a possessive prefix
                possessive_prefix = get_random_possessive_prefix(adjectives_relationships_data_local)
                if possessive_prefix:
                    return f"{possessive_prefix} {relationship_term}"
            return relationship_term
        return None # Fallback if no suitable object term found

    subject_np = generate_subject_noun_phrase(adjectives_relationships_data)

    # Generate an object noun phrase, with a chance of not having one
    object_np = None
    if random.random() < 0.7: # 70% chance to include an object
        max_retries = 3
        retries = 0
        while retries < max_retries:
            temp_object_np = generate_object_noun_phrase(adjectives_relationships_data)
            if temp_object_np is not None and temp_object_np != subject_np:
                object_np = temp_object_np
                break # Found a distinct object, exit loop
            elif temp_object_np is None: # If object generation itself failed
                break # Can't generate a valid object
            retries += 1
        # If loop finishes without finding a distinct object, object_np remains None

    # 1. Choose a random ablaut-triggering particle
    all_particles = engine.ablaut_e_triggers + engine.ablaut_i_triggers
    particle = random.choice(all_particles)

    # 2. Apply ablaut to the verb stem
    modified_verb = engine.apply_vowel_ablaut(verb_stem, particle)

    # 3. Choose one or two random adverbs
    adverbs = []
    num_adverbs = random.randint(0, 2) # Allow 0 adverbs
    for _ in range(num_adverbs):
        adverb = get_random_word(adverbs_data)
        if adverb and adverb not in adverbs: # Avoid duplicates for now
            adverbs.append(adverb)

    # 4. Construct the sentence
    # Lakota word order is generally OSV (Object-Subject-Verb), but for this simplified generator,
    # we'll place the object after the subject and adverbs, before the verb.
    sentence_parts = [subject_np]

    # Insert adverbs
    sentence_parts.extend(adverbs)

    # Insert object if present, before the verb
    if object_np:
        sentence_parts.append(object_np)

    sentence_parts.append(modified_verb)
    sentence_parts.append(particle)

    return " ".join(sentence_parts).capitalize() + "."
```

```python
# Demonstrate automated sentence generation with subject and object noun phrase integration

print("=== Generated Lakota Sentences with Subject and Object Noun Phrases ===\n")

for i in range(7):
    sentence = generate_simple_lakota_sentence(engine, verb_stem="waŋyáŋka")
    print(f"Sentence {i+1}: {sentence}")

print("\n--- Sentences now include dynamically generated subjects and objects. ---")
```
