# Environmental and Social Impacts of Datacenters

```python
import pandas as pd

file_path = '/content/Aterio US Data Centers Dashboard.xlsx'

try:
    df = pd.read_excel(file_path)
    print(f"Successfully loaded data from {file_path}")
    display(df.head())
except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found. Please ensure it is uploaded or the path is correct.")
except Exception as e:
    print(f"Error loading Excel file: {e}")
    print("Please ensure the file is a valid Excel format.")
```

```python
# Filter the DataFrame for 'Active' or 'Construction' stages to ensure filtered_df is available
filtered_df = df[df['DATA_CENTER_STAGE'].isin(['Active', 'Construction'])]

# Calculate the total square footage for the filtered 'Active' and 'Construction' stages
total_square_footage_filtered = filtered_df['s_current_square_footage'].sum()

print(f"Total square footage for 'Active' and 'Construction' data centers: {total_square_footage_filtered:.2f} sq ft")
```

```python
print('DataFrame Info:')
df.info()

print('\nDataFrame Summary Statistics:')
display(df.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Choose a numerical column to visualize, e.g., 's_current_power_capacity_mw'
column_to_visualize = 's_current_power_capacity_mw'

plt.figure(figsize=(10, 6))
sns.histplot(df[column_to_visualize].dropna(), kde=True)
plt.title(f'Distribution of {column_to_visualize}')
plt.xlabel(column_to_visualize)
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# Calculate the average 30-day change in power capacity
average_power_delta = df['s_delta_30d_power_capacity_mw'].mean()
print(f"Average 30-day change in power capacity: {average_power_delta:.2f} MW")

# Calculate the average 30-day change in square footage
average_sq_ft_delta = df['s_delta_30d_square_footage'].mean()
print(f"Average 30-day change in square footage: {average_sq_ft_delta:.2f} sq ft")
```

```python
print('DataFrame Info:')
df.info()

print('\nDataFrame Summary Statistics:')
display(df.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate average power capacity by data center stage
average_power_by_stage = df.groupby('DATA_CENTER_STAGE')['s_current_power_capacity_mw'].mean().reset_index()

# Create the bar chart
plt.figure(figsize=(12, 7))
sns.barplot(x='DATA_CENTER_STAGE', y='s_current_power_capacity_mw', data=average_power_by_stage, palette='viridis')
plt.title('Average Power Capacity (MW) by Data Center Stage')
plt.xlabel('Data Center Stage')
plt.ylabel('Average Power Capacity (MW)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate total power capacity by data center stage
total_power_by_stage = df.groupby('DATA_CENTER_STAGE')['s_current_power_capacity_mw'].sum().reset_index()

# Create the pie chart
plt.figure(figsize=(10, 10))
plt.pie(total_power_by_stage['s_current_power_capacity_mw'],
        labels=total_power_by_stage['DATA_CENTER_STAGE'],
        autopct='%1.1f%%',
        startangle=90,
        pctdistance=0.85) # Adjust pctdistance to place percentages inside slices
plt.title('Distribution of Total Power Capacity by Data Center Stage')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.tight_layout()
plt.show()
```

```python
total_overall_power_capacity = total_power_by_stage['s_current_power_capacity_mw'].sum()
total_power_by_stage['percentage_contribution'] = (total_power_by_stage['s_current_power_capacity_mw'] / total_overall_power_capacity) * 100

print("Percentage Contribution of Each Stage to Total Power Capacity:")
display(total_power_by_stage[['DATA_CENTER_STAGE', 'percentage_contribution']].round(2))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group by DATA_CENTER_STAGE and sum the 30-day deltas for power capacity
power_capacity_delta_by_stage = df.groupby('DATA_CENTER_STAGE')['s_delta_30d_power_capacity_mw'].sum().reset_index()

plt.figure(figsize=(12, 7))
sns.barplot(x='DATA_CENTER_STAGE', y='s_delta_30d_power_capacity_mw', data=power_capacity_delta_by_stage, palette='coolwarm', hue='DATA_CENTER_STAGE', legend=False)
plt.title('30-Day Change in Power Capacity (MW) by Data Center Stage')
plt.xlabel('Data Center Stage')
plt.ylabel('30-Day Change in Power Capacity (MW)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Group by DATA_CENTER_STAGE and sum the 30-day deltas for square footage
square_footage_delta_by_stage = df.groupby('DATA_CENTER_STAGE')['s_delta_30d_square_footage'].sum().reset_index()

plt.figure(figsize=(12, 7))
sns.barplot(x='DATA_CENTER_STAGE', y='s_delta_30d_square_footage', data=square_footage_delta_by_stage, palette='rocket', hue='DATA_CENTER_STAGE', legend=False)
plt.title('30-Day Change in Square Footage by Data Center Stage')
plt.xlabel('Data Center Stage')
plt.ylabel('30-Day Change in Square Footage (sq ft)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## Summary Report: Key US Data Center Metrics

This report summarizes the key metrics and trends observed from the provided US Data Centers Dashboard data. The dataset contains 7 entries and 11 columns, offering insights into various stages of data center development, their power capacities, and square footage.

### Overall Data Overview

The dataset provides information on data centers across different stages, including 'Active', 'Announcement', 'Cancelled', 'Construction', 'Delayed', 'Land Bank', and 'Not Approved/Withdrawn'. Key metrics include current power capacity (MW), current square footage, and 30-day changes for both.

### General Statistics

From the descriptive statistics, we observed a wide range in power capacities and square footage across the different stages. For instance, the maximum current power capacity is 391470.65 MW, indicating a significant size for some facilities.

### Growth Trends

-   **Average 30-day change in power capacity**: The data shows an average increase of **6208.94 MW** in power capacity over a 30-day period.
-   **Average 30-day change in square footage**: Similarly, there was an average increase of **1,933,692.43 sq ft** in square footage over the same period.

### Power Capacity and Stage Analysis

-   **Average Power Capacity by Stage**: 'Announcement' stage data centers have the highest average power capacity, followed by 'Construction'. This suggests a strong pipeline of large-scale projects.
-   **Total Power Capacity Distribution**: 'Announcement' stage data centers contribute significantly to the total power capacity, accounting for **69.04%**. 'Construction' contributes **12.65%**, and 'Active' contributes **9.24%**.

### Square Footage and Stage Analysis

-   The total square footage for 'Active' and 'Construction' data centers combined is **533,532,167.00 sq ft**, highlighting the immense physical footprint of operational and soon-to-be operational facilities.

### 30-Day Delta Analysis by Stage

-   **Power Capacity Delta**: The 'Announcement' stage shows the largest positive 30-day change in power capacity, reinforcing its role in future growth. 'Not Approved/Withdrawn' and 'Active' also show positive changes, while 'Delayed' and 'Land Bank' stages experienced minor negative changes.
-   **Square Footage Delta**: 'Active' and 'Not Approved/Withdrawn' stages show substantial positive 30-day changes in square footage, indicating ongoing expansion and new developments. 'Delayed' and 'Land Bank' stages, however, show significant negative changes, possibly due to projects being reduced or repurposed.

### Conclusion

The US data center market, as represented by this dataset, is characterized by substantial growth, particularly in the 'Announcement' and 'Active' stages. While there are some delays and withdrawals, the overall trend points towards increasing power capacity and physical footprint. The high contribution of 'Announcement' stage data centers suggests a robust pipeline for future expansion.
